```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler

# 加载数据集（不修改原文件）
housing_data = pd.read_csv('housing.csv')

# 显示前几行数据
housing_data.head()
```

```python
# 先排除异常值（不修改文件，仅处理时排除）
housing_data = housing_data[housing_data['median_house_value'] < 500001]

# 处理 total_bedrooms 缺失值（用中位数填补）
housing_data['total_bedrooms'] = housing_data['total_bedrooms'].fillna(housing_data['total_bedrooms'].median())

# 检查数据结构
housing_data.info()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(housing_data['median_house_value'], kde=True, color='skyblue', bins=50)
plt.title('Distribution of Median House Values')
plt.xlabel('Median House Value (USD)')
plt.ylabel('Frequency')
plt.show()
```

```python
# 使用单次编码将分类特征转换为数字特征
housing_data_encoded = pd.get_dummies(housing_data, columns=['ocean_proximity'], drop_first=True)

# 计算数字特征的相关矩阵
corr_matrix = housing_data_encoded.corr()

# 绘制相关矩阵
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

```python
correlation_with_price = corr_matrix['median_house_value'].sort_values(ascending=False)
print(correlation_with_price)
```

```python
plt.figure(figsize=(14, 10))

plt.subplot(1, 2, 1)
sns.scatterplot(x=housing_data['longitude'], y=housing_data['median_house_value'])
plt.title('Longitude vs Median House Value')
plt.xlabel('Longitude')
plt.ylabel('Median House Value')

plt.subplot(1, 2, 2)
sns.scatterplot(x=housing_data['latitude'], y=housing_data['median_house_value'])
plt.title('Latitude vs Median House Value')
plt.xlabel('Latitude')
plt.ylabel('Median House Value')

plt.tight_layout()
plt.show()

# 经纬度热图
plt.figure(figsize=(12, 8))
sns.scatterplot(
    x=housing_data['longitude'], y=housing_data['latitude'],
    hue=housing_data['median_house_value'],
    palette='viridis',
    size=housing_data['median_house_value'],
    sizes=(10, 200), marker='o', legend=None
)
plt.title('Geographic Distribution of House Prices')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

```python
# 数值型特征与目标变量散点图
numerical_columns = ['housing_median_age', 'total_rooms', 'total_bedrooms', 'population', 'households', 'median_income']

plt.figure(figsize=(16, 12))
for i, feature in enumerate(numerical_columns, 1):
    plt.subplot(2, 3, i)
    sns.scatterplot(x=housing_data[feature], y=housing_data['median_house_value'])
    plt.title(f'{feature} vs Median House Value')
    plt.xlabel(feature)
    plt.ylabel('Median House Value')

plt.tight_layout()
plt.show()

# 分类变量 ocean_proximity 对房价的影响
plt.figure(figsize=(8, 6))
sns.boxplot(x=housing_data['ocean_proximity'],  y=housing_data['median_house_value'])
plt.title('Ocean Proximity vs Median House Value')
plt.xlabel('Ocean Proximity')
plt.ylabel('Median House Value')
plt.show()
```
